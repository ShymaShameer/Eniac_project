## Exploring the price details/ Revenue

### Excluding the ouitliers in 2 Approaches:
1. Median Deviation (MAD) method- for removing order_ids where unit_price differ from median ( 50%-200% of median_price)
2. IQR method- for removing order_ids where difference in total paid & total unit price differences are high

#### 1. Median deviation method

```python
orders_df_m = orders_df.copy()
orderlines_df_m = orderlines_df.copy()
```

```python
orderlines_df_m["median_price"] = orderlines_df_m.groupby("sku")["unit_price"].transform("median")
```

```python
orderlines_df_m["price_ratio"] = orderlines_df_m["unit_price"] / orderlines_df_m["median_price"]
```

```python
orderlines_df_m
```

* Keeping the range between 50% -200% of median :
  1.  Since median represents the typical price & is robust to extreme values, so it gives a good estimate of the “normal” price for that product.

  2. Allow normal variation in Prices, which can vary due to discounts,promotions, packaging differences etc. Allowing ±50% -100% variation keeps legitimate prices.

  3. Symmetric tolerance around the typical price 0.5 × median → allows up to 50% cheaper 2 × median → allows up to 100% more expensive

```python
# identifying the outlier order_ids whose price either 2 times higher or 0.5 times lesser than unit price
outlier_order_id =orderlines_df_m.loc[(orderlines_df_m["price_ratio"] > 2) | (orderlines_df_m["price_ratio"] < 0.5),"order_id"].unique()
outlier_order_id
```

```python
# Updating orders & orderline dataframes by removing the outlier order ids in the list (outlier_order_id )
```

```python
orders_df_m= orders_df_m.loc[~ orders_df_m["order_id"].isin(outlier_order_id),:]
```

```python
orderlines_df_m= orderlines_df_m.loc[~ orderlines_df_m["order_id"].isin(outlier_order_id),:]
```

```python
# checking if both dataframes have same ids
orders_df_m["order_id"].nunique(), orderlines_df_m["order_id"].nunique()
```

#### 2. IQR method

```python
# creating new column with unit_price total
orderlines_df_m["unit_price_total"] = (orderlines_df_m["unit_price"])* (orderlines_df_m["product_quantity"])

# filtering to view only required columns for price comparison
orderlines_pc= orderlines_df_m[["order_id", "product_quantity","sku", "unit_price","unit_price_total"]]
orderlines_pc.sample(10)
```

```python
# Groupingby order_id & aggregating by unit_price_total to find the total for each order
orderlines_order_ids_m = orderlines_pc.groupby("order_id",as_index=False)["unit_price_total"].sum()
orderlines_order_ids_m
```

```python
diff_df_m= orders_df_m.merge(orderlines_order_ids_m,on ="order_id")
diff_df_m["difference"]= (diff_df_m["total_paid"]) - (diff_df_m["unit_price_total"])
diff_df_m
```

```python
diff_df_m.loc[diff_df_m["difference"] < 0,:]. sort_values("difference",ascending= True).head(20)
```

```python
diff_df_m.describe().round(2)
```

```python
# calculating the quartiles
Q1 = diff_df_m["difference"].quantile(0.25)
Q3 = diff_df_m["difference"].quantile(0.75)

Q1, Q3
```

```python
# calculating the interquartile range
IQR = Q3-Q1
# lower boundary
L= Q1 - 1.5*IQR
# upper boundary
U= Q3 + 1.5*IQR
print(f" Lower boundary: {L}\n Upper boundary: {U}")
```

```python
# filter the DataFrame to include only "non-outliers"
diff_no_outliers_df = diff_df_m.loc[
    (diff_df_m["difference"] >= (L))
    &
    (diff_df_m["difference"] <= (U))
    ,
    :]
```

```python
normal_diff_list = diff_no_outliers_df["order_id"]
```

```python
orders_df_m = orders_df_m.loc[orders_df_m["order_id"].isin(normal_diff_list), :]
orderlines_df_m = orderlines_df_m.loc[orderlines_df_m["order_id"].isin(normal_diff_list), :]
```
